# Dropout Layers
When large feedforward neural networks are trained on small datasets (datasets which contain datapoints smaller in number compared to the total trainable parameters of the model) the model often overfits. Even if **L1** or **L2** normalization is applied while training the shear number of trainable parameters do not prevent models from over fitting. The reason the model with large trainable parameters overfit is that they learn the noise present in the dataset along with the actual information.

One way to handle this over fitting issue is by averaging the results of many smaller neural networks. This is computationally expensive to do during training and testing time. To perform this same action of averaging the results of many smaller neural networks but avoid training multiple neural networks is to introduce a **dropout layer**.

The authors of the paper [Improving neural networks by preventing co-adaptation of feature detectors](https://arxiv.org/pdf/1207.0580) suggest to use a **dropout layer** as part of a neural network. In a dropout layer in every **forward** pass during **training time** output from different neurons present in the incoming layer are randomly dropped with a probability of **p**. This **p** is the hyper parameters that controls how many neurons will be active at any given time during the training phase. This prevents neurons to only predict the right result when other neurons achieve a specific value (this is called complex co adaptation).

As large and different combinations of neurons are formed due to a dropout layer in each forward pass, every neuron get's a chance to learn and detect features and they do not have to rely on other neurons to achieve a specific value in order to do so. Do note that dropout leads to slower convergence rate when compared to training without dropouts layers but the end result of training with dropout will always lead to a better result during inference or test time.

## How to implement a Dropout Layer
A dropout layer behaves in different ways during the training and testing phase. 

### During training phase
The dropout layers drop the outputs of the previous layer with a probability of **p**. What this means is that the values are set to **0** which would result in no information getting propagating from the neurons that have been dropped. This forces neurons to learn the features in the input in a different combination every forward pass effectively acting as multiple smaller neural networks being trained on the dataset simultaneously.

### During testing phase
The neurons in the testing phase do not get dropped but since more number of neurons are participating during the test phase when compared to training phase we need to scale the output to match the expected output seen during the training phase so we scale them by a factor of $\frac{1}{1 - p}$. This scaling approximates averaged ensemble prediction of the multiple smaller sub-networks sampled during training.

## A trick done in pytorch implmenentation
Based on the above understanding we see that there are two steps

Step 1: Drop neurons from previous layer with a probability of **p** during training phase. <br>
Step 2: During testing phase scale the outputs of the previous layer by a factor of $\frac{1}{1 - p}$ to scale to the expected output seen during the training phase.

What is done in the pytorch's implementation is that during the training phasee you drop the neurons with a probability of **p** and for the remaining outputs we scale them by the factor $\frac{1}{1 - p}$ and during the testing phase we just return the incoming values as they are. This eliminates the step of scaling the output by the factor.
