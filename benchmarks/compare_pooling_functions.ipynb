# Compare custom implementation of the pooling functions with pytorch's

This notebook will show comparison between how the custom implementation of pooling function fairs against torch's implementation

```python
import torch
from pytorch_fundamentals.layers.pooling import MaxPool2dCustom, AvgPool2dCustom
```

## Reusable pieces of code

code written in the very next cell will be used through out the notebook

```python
# Tensor to be used throughout for comparing custom pooling layer implementations with that of pytorch
X = torch.arange(0, 27, dtype=torch.float32).reshape((1, 3, 3, 3))
# Convolution filter hyper parameters
kernel_size = 2
stride = 1
padding = 0
dilation = 1

# Function to be used to compare custom implementations with that of pytorch
def compare_max_pool_implementation(X: torch.Tensor,
                                    custom_implementation_class: torch.nn.Module, 
                                    pytorch_class: torch.nn.Module,
                                    kernel_size: int,
                                    stride: int,
                                    padding: int,
                                    dilation: int) -> None:
    """
    Function that generates a comparision between custom implementation of max pooling functions with that of pytorch'

    :param X: Input tensor to base the comparison on
    :param custom_implementationc_class: Custom class that implements the pooling function
    :param pytorch_class: Pytorch's implementation

    :returns: None
    """

    # Instantiate custom and pytorch pooling classes
    custom_instance: torch.nn.Module = custom_implementation_class(kernel_size=kernel_size,
                                                                   stride=stride,
                                                                   padding=padding,
                                                                   dilation=dilation)
    pytorch_instance: torch.nn.Module = pytorch_class(kernel_size=kernel_size,
                                                      stride=stride,
                                                      padding=padding,
                                                      dilation=dilation)

    # Generate the output after forward propagation of X through the functions
    output_custom: torch.Tensor = custom_instance.forward(X)
    output_pytorch: torch.Tensor = pytorch_instance.forward(X)

    # Compare the functions outputs and provide the answer
    if torch.all(torch.isclose(output_custom, output_pytorch)):
        print("The implementation is correct")
    else:
        print("The implementation is incorrect")


def compare_avg_pool_implementation(X: torch.Tensor,
                                    custom_implementation_class: torch.nn.Module, 
                                    pytorch_class: torch.nn.Module,
                                    kernel_size: int,
                                    stride: int,
                                    padding: int) -> None:
    """
    Function that generates a comparision between custom implementation of average pooling functions with that of pytorch'

    :param X: Input tensor to base the comparison on
    :param custom_implementationc_class: Custom class that implements the pooling function
    :param pytorch_class: Pytorch's implementation

    :returns: None
    """

    # Instantiate custom and pytorch pooling classes
    custom_instance: torch.nn.Module = custom_implementation_class(kernel_size=kernel_size,
                                                                   stride=stride,
                                                                   padding=padding)
    pytorch_instance: torch.nn.Module = pytorch_class(kernel_size=kernel_size,
                                                      stride=stride,
                                                      padding=padding)

    # Generate the output after forward propagation of X through the functions
    output_custom: torch.Tensor = custom_instance.forward(X)
    output_pytorch: torch.Tensor = pytorch_instance.forward(X)

    # Compare the functions outputs and provide the answer
    if torch.all(torch.isclose(output_custom, output_pytorch)):
        print("The implementation is correct")
    else:
        print("The implementation is incorrect")
```

# Max Pooling 2 Dimensional Layer

```python
compare_max_pool_implementation(X, 
                                custom_implementation_class=MaxPool2dCustom, 
                                pytorch_class=torch.nn.MaxPool2d,
                                kernel_size=kernel_size,
                                stride=stride,
                                padding=padding,
                                dilation=dilation)
```

# Average Pooling 2 Dimensional Layer

```python
compare_avg_pool_implementation(X, 
                                custom_implementation_class=AvgPool2dCustom, 
                                pytorch_class=torch.nn.AvgPool2d,
                                kernel_size=kernel_size,
                                stride=stride,
                                padding=padding)
```
