# Compare custom implementation of normalization functions with pytorch's

This notebook will show comparison between how the custom implementation of normalization function fairs against torch's implementation

```python
import torch
from pytorch_fundamentals.layers.normalization import BatchNorm2dCustom
```

## Reusable pieces of code

code written in the very next cell will be used through out the notebook

```python
# Tensor to be used throughout for comparing custom 2D batch normalization layer implementation with that of pytorch
X = torch.arange(0, 27, dtype=torch.float32).reshape((1, 3, 3, 3))

eps = 0.0001
momentum = 0.1
num_features = 3

# Function to be used to compare custom implementations with that of pytorch
def compare_batch_normalization_implementation(X: torch.Tensor,
                                               custom_implementation_class: torch.nn.Module, 
                                               pytorch_class: torch.nn.Module,
                                               num_features: int,
                                               momentum: float,
                                               eps: float) -> None:
    """
    Function that generates a comparision between custom implementation of batch normalization functions with that of pytorch'

    :param X: Input tensor to base the comparison on
    :param custom_implementationc_class: Custom class that implements the pooling function
    :param pytorch_class: Pytorch's implementation
    :param num_features: Number of channels present in the 4 D tensor to normalize for 
    :param momentum: Momentum to be used to update running mean and variance
    :param eps: Value to avoid computational overflow.
 
    :returns: None
    """

    # Instantiate custom and pytorch pooling classes
    custom_instance: torch.nn.Module = custom_implementation_class(num_features=num_features, momentum=momentum, eps=eps)
    pytorch_instance: torch.nn.Module = pytorch_class(num_features=num_features, momentum=momentum, eps=eps)

    # Generate the output after forward propagation of X through the functions
    output_custom: torch.Tensor = custom_instance.forward(X)
    output_pytorch: torch.Tensor = pytorch_instance.forward(X)

    # Compare the functions outputs and provide the answer
    if torch.all(torch.isclose(output_custom, output_pytorch, atol=1e-6)):
        print("The implementation is correct")
    else:
        print("The implementation is incorrect")
```

## 2 Dimensional Batch Normalization Layer

```python
compare_batch_normalization_implementation(X,
                                           BatchNorm2dCustom, 
                                           torch.nn.BatchNorm2d,
                                           num_features=num_features,
                                           momentum=momentum,
                                           eps=eps)
```
